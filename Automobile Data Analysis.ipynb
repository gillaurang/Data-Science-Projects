```python
# Your code (F3) Three scatter plots side by side using `plt.subplots(1, 3)`: mileage against
# horsepower, against wheelbase, and against length.

print(df.plot.scatter(x='averagemileage', y='horsepower', color='green'))

print(df.plot.scatter(x='averagemileage', y='wheelbase', color='indianred'))

print(df.plot.scatter(x='averagemileage', y='length', color='indianred'))
```

```python
# Your code (F4)  A histogram comparing the price of German cars with the price of Japanese cars on
# the same axes, with a legend.
df[df['Country'] == 'Germany']['price'].hist(alpha=0.8, label='German Cars', color='indianred')
df[df['Country'] == 'Japan']['price'].hist(alpha=0.8, label='Japanese Cars', color='skyblue')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Price Distribution of German and Japanese Cars')
plt.legend()
plt.show() 
```

**Your answer (F2):** as horsepower rises, mileage __deacreases___ . The relationship looks ____strong negative __

**Your answer (F3):** all three plots slope the same way. Give one explanation, other than
"powerful engines burn fuel", for why a car's **length** would also be related to its mileage. ___Length and horsepower are themselves correlated.___

**Your answer (F4):** the two histograms overlap between roughly \$___8000___ and \$_17000_____

```python
cols = ['wheelbase', 'length', 'horsepower', 'averagemileage', 'price']
corr = df[cols].corr()

fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)

ax.set_xticks(range(len(cols)))
ax.set_xticklabels(cols, rotation=45, ha='right')
ax.set_yticks(range(len(cols)))
ax.set_yticklabels(cols)

# write the number in each cell
for i in range(len(cols)):
    for j in range(len(cols)):
        ax.text(j, i, f'{corr.iloc[i, j]:.2f}', ha='center', va='center', color='black')

fig.colorbar(im, label='Pearson correlation')
ax.set_title('Correlation between numeric features')
plt.tight_layout()
plt.show()
```
